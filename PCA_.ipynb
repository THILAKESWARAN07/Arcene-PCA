```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, classification_report
from sklearn.ensemble import RandomForestClassifier
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('arcene_train.data', delim_whitespace=True, header=None)
train_labels =pd.read_csv('arcene_train.labels',delim_whitespace=True, header=None)
data_valid = pd.read_csv('arcene_valid.data', delim_whitespace=True, header=None)
labels_valid = pd.read_csv('arcene_valid.labels', delim_whitespace=True, header=None)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
data_valid_scaled = scaler.transform(data_valid)
```

```python
# 3. PCA to reduce to 2 principal components for visualization
pca = PCA(n_components=0.80)

X_pca = pca.fit_transform(X_scaled)
data_valid_pca = pca.transform(data_valid_scaled)
```

```python
print(f"Reduced shape:  {X_pca.shape}")
```

```python
train_labels.value_counts()
```

```python
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_pca, train_labels)

print("Model training complete.")

print("Evaluating model...")

y_pred = model.predict(data_valid_pca)

accuracy = accuracy_score(labels_valid, y_pred)
print("\n--- 🎯 Results ---")
print(f"Model Accuracy on Test Data: {accuracy :.2f}%")

print("\nClassification Report:")
print(classification_report(labels_valid, y_pred))
```
